## WRF mean std

```python
config_name = os.path.realpath('data_config_WRF.yml')

with open(config_name, 'r') as stream:
    conf = yaml.safe_load(stream)
```

```python
# ds = xr.open_dataset('/glade/derecho/scratch/ksha/DWC_data/CONUS_domain_FULL/C404/C404_FULL_1986.zarr')
```

```python
varnames = list(conf['zscore'].keys())
varnames = varnames[:-3] # remove save_loc and others

varname_upper = ['WRF_U', 'WRF_V', 'WRF_T', 'WRF_Q', 'WRF_P', 'WRF_W', 'WRF_Q_tot', 'WRF_Q_tot_05', 'WRF_Z'] 
varname_surf = list(set(varnames) - set(varname_upper))

N_level = 12
```

```python
for i_var, varname in enumerate(varname_surf):
    
    f = open('{}zscore_WRF_{}.sh'.format(conf['qsub']['qsub_loc'], varname), 'w') 
    
    heads = '''#!/bin/bash -l

#PBS -N {}
#PBS -A P48500028
#PBS -l walltime=23:59:59
#PBS -l select=1:ncpus=16:mem=128GB
#PBS -q casper
#PBS -o zscore_WRF_{}.log
#PBS -e zscore_WRF_{}.err

conda activate credit
cd {}
python zscore_WRF_var.py {} {}
'''.format(varname, varname, varname, 
           conf['qsub']['scripts_loc'], 
           varname,
           'None')
    
    print(heads, file=f)    
    f.close()

f = open('{}preprocess_surf_zscore_WRF.sh'.format(conf['qsub']['qsub_loc']), 'w')

for i_var, varname in enumerate(varname_surf):
    print('qsub zscore_WRF_{}.sh'.format(varname), file=f)
    
f.close()
```

```python
for i_var, varname in enumerate(varname_upper):
    for i_level in range(N_level):
        
        f = open('{}zscore_WRF_{}_level{}.sh'.format(conf['qsub']['qsub_loc'], varname, i_level), 'w') 
        
        heads = '''#!/bin/bash -l
    
#PBS -N {}
#PBS -A P48500028
#PBS -l walltime=23:59:59
#PBS -l select=1:ncpus=16:mem=128GB
#PBS -q casper
#PBS -o zscore_WRF_{}.log
#PBS -e zscore_WRF_{}.err

conda activate credit
cd {}
python zscore_WRF_var.py {} {}
'''.format(varname, varname, varname, 
           conf['qsub']['scripts_loc'], 
           varname,
           i_level)
        
        print(heads, file=f)    
        f.close()

f = open('{}preprocess_upper_zscore_WRF.sh'.format(conf['qsub']['qsub_loc']), 'w')

for i_var, varname in enumerate(varname_upper):
    for i_level in range(N_level):
        print('qsub zscore_WRF_{}_level{}.sh'.format(varname, i_level), file=f)
    
f.close()
```
